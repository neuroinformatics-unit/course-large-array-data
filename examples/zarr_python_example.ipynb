# Zarr-python

## Zarr arrays

Zarr arrays contain our n-dimensional array data

```python
import zarr
```

```python
# Create a zarr array at data/example-1.zarr

zarr_array = zarr.create_array(
  store="data/example-1.zarr", 
  shape=(1000, 1000),  # overall size of the array
  chunks=(100, 100),  # size of each chunk
  dtype='uint8'
)
zarr_array[:] = 42
```

If you go to data/example1.zarr in your filebrowser, you'll see it contains:
- a folder called c
- a zarr.json file

The c folder contains all the chunk files for this array. `zarr.json` contains basic metadata about the array:
- shape
- chunk size
- data type...

## Zarr groups

Zarr groups allow us to group together multiple arrays

```python
group_1 = zarr.create_group(store="data/group-1")
```

```python
# Create two arrays inside group_1
for array_name in ["array_1", "array_2"]:
    arr = group_1.create_array(
        name=array_name,
        shape=(1000, 1000),
        chunks=(100, 100),
        dtype='uint8'
    )
    arr[:] = 42
```

If you go to `data/group-1` in your filebrowser, you should see this folder contains two sub-folders (`array_1` and `array_2`) as well as a `zarr.json` file.

In fact, Zarr allows us to put groups inside of other groups. We can build up a tree of groups and arrays, as complex as we like.

```python
# Make a group inside our group
group_2 = group_1.create_group("group_2")
```

```python
# Add an array to the sub-group
arr_3 = group_2.create_array(
    name="array_3",
    shape=(1000, 1000),
    chunks=(100, 100),
    dtype='uint8'
)
arr_3[:] = 42
```

```python
# Print out the whole hierarchy
print(group_1.tree())
```
