# Viewing OME-Zarr data

Let's try using `napari` to visualise images stored in `OME-Zarr`. First, let's open one of the datasets we created in an earlier notebook:

```python
import napari

viewer = napari.Viewer()
viewer.open("data/exported-heart-multiscale.ome.zarr")

# start the event loop and show the viewer
napari.run()
```

```python
# It has recognised our multiple levels and loaded as 'multi-scale data'
print(viewer.layers[0].data)
```

Napari will dynamically select a resolution level based on how zoomed in the viewer is. In the 'layer controls' menu, you should see a 'resolution' dropdown that is set to 'Auto' by default.

Try setting this drop-down to different set resolution levels - what changes in the viewer?

## Key take-aways
